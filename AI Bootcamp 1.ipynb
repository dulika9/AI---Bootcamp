```python
import random

numbers = [random.randint(1, 100) for i in range(20)]

print("List of random numbers:")
print(numbers)

print("Even numbers:")
for num in numbers:
    if num % 2 == 0:
        print(num)
```

```python
text = "Artificial Intelligence is interesting"

# Reverse the complete string
print("Reverse of string:")
print(text[::-1])

# Reverse the order of words
print("Reverse order of words:")
print(" ".join(text.split()[::-1]))
```
