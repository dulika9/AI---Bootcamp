```python
import pandas as pd
from scipy import stats

# Create the dataset
data = {
    "Student": ["A", "B", "C", "D", "E"],
    "Math": [90, 78, 85, 60, 95],
    "Science": [85, 92, 78, 65, 88],
    "English": [88, 80, 90, 70, 92]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

# Minimum
print("\nMinimum:")
print(df[["Math", "Science", "English"]].min())

# Maximum
print("\nMaximum:")
print(df[["Math", "Science", "English"]].max())

# Mean
print("\nMean:")
print(df[["Math", "Science", "English"]].mean())

# Median
print("\nMedian:")
print(df[["Math", "Science", "English"]].median())

# Mode
print("\nMode:")
print(df[["Math", "Science", "English"]].mode())

# Standard deviation
print("\nStandard Deviation:")
print(df[["Math", "Science", "English"]].std())
```

```python
# Step 1: Importing NumPy
import numpy as np

# Print a welcome message
print("Welcome to NumPy - The Foundation of AI and Data Analysis!")

x = np.random.randint(1,1000,size=(4000,1000))
x

# Create a dataset: Runs scored by 5 players in 3 matches
players = np.array(['Sachin', 'Virat Kohli', 'Yashasvi', 'Jaspreet', 'Rishab'])
matches = np.array(['IPL 2015', 'IPL 2020', 'IPL 2023'])

# Randomly generate runs scored by players
#np.random.seed(42)  # For reproducibility
runs = np.random.randint(10, 100, size=(len(players), len(matches)))

# Print the dataset
print("Players:", players)
print("Matches:", matches)
print("Runs scored by each player:")
print(runs)

# Calculate total runs scored by each player
total_runs_by_player = runs.sum(axis=1)
print("Total runs scored by each player:")
for player, total in zip(players, total_runs_by_player):
    print(f"{player}: {total} runs")
    
# Calculate average runs per match
average_runs_per_match = runs.mean(axis=0)
print("Average runs scored in each match:")
for match, avg in zip(matches, average_runs_per_match):
    print(f"{match}: {avg:.2f} runs")
    
    
# Find the player with the highest total runs
most_runs_index = np.argmax(total_runs_by_player)
print(f"The player with the most runs is {players[most_runs_index]} with {total_runs_by_player[most_runs_index]} runs.")

# Calculate the standard deviation of runs for each player
std_dev_by_player = runs.std(axis=1)
print("Variability in performance (Standard Deviation):")
for player, std in zip(players, std_dev_by_player):
    print(f"{player}: {std:.2f}")
    
# Import matplotlib for plotting
import matplotlib.pyplot as plt

# Bar chart of total runs by player
plt.figure(figsize=(8, 5))
plt.bar(players, total_runs_by_player, color='skyblue')
plt.title("Total Runs Scored by Each Player")
plt.xlabel("Players")
plt.ylabel("Total Runs")
plt.show()


    
```

```python
# Create a dataset: Olympic medals for 5 countries over 4 years
countries = np.array(['USA', 'China', 'India', 'Germany', 'Australia'])
years = np.array([2012, 2016, 2020, 2024])  # Olympic years

# Randomly generate the number of medals (Gold, Silver, Bronze) for each country in each year
np.random.seed(42)  # For reproducibility
medals = np.random.randint(10, 100, size=(len(countries), len(years), 3))  # 3 types of medals

# Print the dataset
print("Countries:", countries)
print("Years:", years)
print("Medals (Gold, Silver, Bronze):")
print(medals)

# Total medals won by each country
total_medals_by_country = medals.sum(axis=2)  # Sum across medal types
print("Total medals by each country over the years:")
for country, total in zip(countries, total_medals_by_country):
    print(f"{country}: {total}")
    
# Which country won the most medals in 2020?
medals_2020 = total_medals_by_country[:, 2]  # Medals in the 3rd year (2020)
most_medals_index = np.argmax(medals_2020)
print(f"Country with the most medals in 2020: {countries[most_medals_index]} ({medals_2020[most_medals_index]} medals)")

# Average medals by year
average_medals_by_year = total_medals_by_country.mean(axis=0)
print("Average medals won by all countries in each year:")
for year, avg in zip(years, average_medals_by_year):
    print(f"{year}: {avg:.2f}")
    
# Standard deviation of medals (measure of variability)
std_dev_by_year = total_medals_by_country.std(axis=0)
print("Standard deviation of medals won by all countries in each year:")
for year, std in zip(years, std_dev_by_year):
    print(f"{year}: {std:.2f}")
    
import matplotlib.pyplot as plt

# Plot total medals for each country
plt.figure(figsize=(10, 6))
for i, country in enumerate(countries):
    plt.plot(years, total_medals_by_country[i], label=country, marker='o')

plt.title('Total Olympic Medals by Country Over Years')
plt.xlabel('Year')
plt.ylabel('Total Medals')
plt.legend()
plt.grid(True)
plt.show()    
    
    
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Generate engineering data using NumPy
time = np.linspace(0, 10, 500)  # 500 points from 0 to 10 seconds
amplitude1 = np.sin(2 * np.pi * time)  # Simple sine wave
amplitude2 = np.exp(-0.5 * time) * np.sin(2 * np.pi * time)  # Damped sine wave

# 2. Create a figure with 2 subplots vertically (2 rows, 1 column)
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 8), dpi=100)

# 3. Plot 1: Simple Sine Wave (Top Subplot)
ax1.plot(
    time,
    amplitude1,
    color="crimson",
    linewidth=2,
    label="Undamped Harmonic Motion",
)
ax1.set_title(
    "Harmonic Response of Structure", fontsize=14, fontweight="bold", pad=12
)
ax1.set_xlabel("Time (seconds)", fontsize=11)
ax1.set_ylabel("Displacement (mm)", fontsize=11)
ax1.grid(True, linestyle="--", alpha=0.6)
ax1.legend(loc="upper right", frameon=True, shadow=True)

# 4. Plot 2: Damped Sine Wave (Bottom Subplot)
ax2.plot(
    time, amplitude2, color="navy", linewidth=2, label="Damped Harmonic Motion"
)
ax2.set_title(
    "Damped Harmonic Response of Structure", fontsize=14, fontweight="bold", pad=12
)
ax2.set_xlabel("Time (seconds)", fontsize=11)
ax2.set_ylabel("Displacement (mm)", fontsize=11)
ax2.grid(True, linestyle="--", alpha=0.6)
ax2.legend(loc="upper right", frameon=True, shadow=True)

# 5. Adjust layout to avoid overlapping labels and display
plt.tight_layout()
# plt.savefig('sensor_data_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd

# Sample messy dataset
data = {
    'Name': ['  Alice  ', 'Bob123', 'Charlie!!', 'David_99', 'Eve@Home'],
    'Phone': ['(123) 456-7890', '123.456.7890', '123-456-7890', '1234567890', 'N/A'],
    'Email': ['alice@mail.com', 'bob@@mail..com', 'charlie@mail.com', 'invalid_email', 'eve@mail.com']
}
df = pd.DataFrame(data)
print("Original Data:")
print(df)

# Clean 'Name': strip whitespace and remove non-alphabetic characters
df['Name_Clean'] = df['Name'].str.strip().str.replace(r'[^a-zA-Z\s]', '', regex=True)

# Clean 'Phone': keep only digits using regex
df['Phone_Clean'] = df['Phone'].str.replace(r'\D', '', regex=True)

# Validate 'Email' using a regex pattern
email_pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'
df['Email_Valid'] = df['Email'].str.match(email_pattern)

print("\nCleaned Data:")
print(df)
```
