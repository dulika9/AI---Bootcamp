```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Paragraph about Drawing
text = """
Drawing is a creative form of visual expression that allows people to communicate
their ideas, thoughts, and imagination through lines, shapes, and patterns.
It is an important skill used in art, design, education, engineering, and many
other fields. Drawing can be done using traditional tools such as pencils,
pens, and colors, or with digital tools and software. Regular drawing helps
improve creativity, concentration, observation, and problem-solving skills.
It also provides a way to express emotions and represent objects, people,
nature, and imaginary ideas visually. In the modern world, drawing is widely
used in graphic design, animation, architecture, illustration, and digital art.
"""

# Create the word cloud
wordcloud = WordCloud(
    width=1000,
    height=600,
    background_color="white",
    max_words=100
).generate(text)

# Display the word cloud
plt.figure(figsize=(12, 7))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud on Drawing")
plt.show()
```
